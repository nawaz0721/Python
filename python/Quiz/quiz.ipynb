### Scenario 1: File Processing and Exception Handling

```python
class GradeError(Exception):
    pass

def calculate_average(filename):
    try:
        grades = []
        with open(filename) as f:
            for line in f:
                name, grade = line.strip().split(',')
                if not grade.strip().isdigit():
                    raise GradeError(f"Invalid grade: {name} - {grade}")
                grades.append(float(grade))
        
        avg = sum(grades) / len(grades)
        print(f"Average Grade: {avg:.2f}")
    
    except FileNotFoundError:
        print("Error: File not found")
    except GradeError as e:
        print(f"Error: {e}")

#run the program
calculate_average("student_records.txt")
```

### Scenario 2: Object-Oriented Programming and Inheritance

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
    
    def info(self):
        return f"{self.year} {self.make} {self.model}"

class Car(Vehicle):
    def __init__(self, make, model, year, doors):
        super().__init__(make, model, year)
        self.doors = doors
    
    def info(self):
        return f"{super().info()} - {self.doors} doors"

class Truck(Vehicle):
    def __init__(self, make, model, year, capacity):
        super().__init__(make, model, year)
        self.capacity = capacity
    
    def info(self):
        return f"{super().info()} - {self.capacity} tons"

# Test
car = Car("Toyota", "Corolla", 2022, 4)
truck = Truck("Ford", "F150", 2020, 2.5)
print(car.info())
print(truck.info())
```

### Scenario 3: Functions and Modules

```python
class Library:
    def __init__(self):
        self.books = {}
        self.borrowed = {}
    
    def add_book(self, id, title, qty=1):
        self.books[id] = self.books.get(id, 0) + qty
    
    def borrow_book(self, id, student):
        if self.books.get(id, 0) > 0:
            self.books[id] -= 1
            self.borrowed[id] = student
            print(f"{student} borrowed book {id}")
        else:
            print("Book not available")
    
    def return_book(self, id):
        if id in self.borrowed:
            self.books[id] = self.books.get(id, 0) + 1
            print(f"Book {id} returned")
        else:
            print("Book was not borrowed")

# Usage
lib = Library()
lib.add_book("B001", "Python", 3)
lib.borrow_book("B001", "Ahmed Nawaz Gondal")
lib.return_book("B001")
```

### Scenario 4: Debugging Code

```python
def calculate(numbers):
    total = sum(numbers)
    product = 1
    for n in numbers:
        product *= n
    average = total / len(numbers)
    return total, product, average

nums = [1, 2, 3, 4, 5]
t, p, a = calculate(nums)
print(f"Total: {t}, Product: {p}, Average: {a}")
```
