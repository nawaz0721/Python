9. A shop will give discount of 10% if the cost of purchased quantity is more than 1000. Ask user for quantity Suppose, one unit will cost 100. Judge and print total cost for user.

```python
print("Shop Discount Program")

quantity = int(input("Enter quantity: "))
unit_price = 100

total_cost = quantity * unit_price

if total_cost > 1000:
   discount = (total_cost * 10)/100
   total_cost = total_cost - discount
   print("Your total cost is", total_cost,"so you got 10% discount")
else:
    print("Your total cost is", total_cost,"so you can not get discount")
```

11. A student will not be allowed to sit in exam if his/her attendence is less than 75%.

Take following input from user  
Number of classes held  
Number of classes attended.  
And print   
percentage of class attended   

Is student is allowed to sit in exam or not.

```python
print("Student allowed to sit in exam or not")

numberOfClassesHeld = int(input("Number of Classes Held"))
numberOfClassesAttended = int(input("Number of Classes Attended"))

totalClass = (numberOfClassesAttended*100)/numberOfClassesHeld
print(totalClass,"% Attendence")
if totalClass >= 75:
    print("Student is Allowed to sit in Exam")
else:
    print("Student is not Allowed to sit in Exam")
```

13. Write a program to check if a year is leap year or not.
If a year is divisible by 4 then it is leap year but if the year is century year like 2000, 1900, 2100 then it must be divisible by 400.

```python
print("Check if a year is leap year or not")

year = int(input("Enter a year:"))

if year % 400 == 0:
    print(year, "is a Leap Year")
elif year % 100 == 0:
    print(year, "is NOT a Leap Year")
elif year % 4 == 0:
    print(year, "is a Leap Year")
else:
    print(year, "is NOT a Leap Year")
```

16. Take input of age of 3 people by user and determine oldest and youngest among them.

```python
print("Find the Oldest and Youngest Person")

age1 = int(input("Enter age of first person: "))
age2 = int(input("Enter age of second person: "))
age3 = int(input("Enter age of third person: "))

oldest = max(age1, age2, age3)
youngest = min(age1, age2, age3)

print("Oldest age is:", oldest)
print("Youngest age is:", youngest)
```
