### Loop

1- Print numbers from 1 to 10 using for loop

```python
for i in range(1, 11):
    print(i)
```

2- Print numbers from 20 to 1 using while loop

```python
i = 20
while i >= 1:
    print(i)
    i -= 1
```

3- Print even numbers from 1 to 10

```python
for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```

4- Print numbers from 1 to n

```python
n = int(input("Enter a number: "))
for i in range(1, n + 1):
    print(i)
```

5- Print odd numbers between 1 and n

```python
n = int(input("Enter a number: "))
for i in range(1, n + 1):
    if i % 2 != 0:
        print(i)
```

6- Print 'Happy Birthday!' five times

```python
for i in range(5):
    print("Happy Birthday!")
```

7- Generate squares of natural numbers

```python
n = int(input("Enter a number: "))
print(f"The first {n} terms of the series are:")
for i in range(1, n + 1):
    print(i * i, end=" ")
```

8- Multiplication table

```python
num = int(input("Enter a number: "))
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

9- Arithmetic progression

```python
start = 16
diff = 61
print("The sequence is:")
for i in range(8):
    print(start + i * diff, end=" ")
```

10- Geometric sequence

```python
start = 12
ratio = 33
print("The sequence is:")
current = start
for i in range(6):
    print(current, end=" ")
    current *= ratio
```

11- Sum of integers from 1 to n

```python
n = int(input("Enter a positive integer: "))
total = 0
for i in range(1, n + 1):
    total += i
print(f"The sum of numbers from 1 to {n} is: {total}")
```

12- Sum of reciprocals

```python
n = int(input("Enter a positive integer: "))
total = 0
for i in range(1, n + 1):
    total += 1 / i
print(f"The sum of reciprocals from 1 to {n} is: {total:.2f}")
```

13- Running total of 5 numbers

```python
total = 0
for i in range(5):
    num = int(input("Enter a number: "))
    total += num
print(f"The final running total is: {total}")
```

14- Factorial calculation

```python
num = int(input("Enter a positive integer: "))
if num < 0:
    print("Factorial does not exist for negative numbers")
elif num == 0:
    print("The factorial of 0 is 1")
else:
    factorial = 1
    for i in range(1, num + 1):
        factorial *= i
    print(f"The factorial of {num} is {factorial}")
```

15-  Power calculation without ** or math.pow()

```python
base = float(input("Enter base number: "))
exponent = int(input("Enter exponent: "))

if exponent == 0:
    result = 1
elif exponent > 0:
    result = 1
    for i in range(exponent):
        result *= base
else: 
    result = 1
    for i in range(-exponent):
        result *= base
    result = 1 / result

print(f"{base} raised to the power {exponent} is: {result}")
```
