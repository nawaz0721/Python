### String Assignment

1. Write a program that accepts a string from user. Your program should count and display number of vowels in that string

```python
print("Count Vowels in a String")

text = input("Enter a string: ")

vowels = "aeiouAEIOU"
count = 0

for char in text:
    if char in vowels:
        count += 1

print("Number of vowels in the string:", count)
```

2. Write a program that reads a string from the keyboard and displays:

- The number of uppercase letters in the string
- The number of lowercase letters in the string
- The number of digits in the string
- The number of whitespace characters in the string

```python
string = input("Enter a string: ")
uppercase = 0
lowercase = 0
digits = 0
whitespace = 0
for char in string:
    if char.isupper():
        uppercase += 1
    elif char.islower():
        lowercase += 1
    elif char.isdigit():
        digits += 1
    elif char.isspace():
        whitespace += 1
print(f"Uppercase letters: {uppercase}")
print(f"Lowercase letters: {lowercase}")
print(f"Digits: {digits}")
print(f"Whitespace characters: {whitespace}")
```

3. Write a Python program that accepts a string from the user. Your program should create and display a new string where the first and last characters have been exchanged.

```python
string = input("Enter a string: ")
if len(string) > 1:
    new_string = string[-1] + string[1:-1] + string[0]
    print("New string:", new_string)
else:
    print("String too short to swap!")
```

4. Write a Python program that accepts a string from the user. Your program should create a new string that is the reverse of the first string and display it.

```python
string = input("Enter a string: ")
new_string = string[::-1]
print("Reversed string:", new_string)
```

5. Write a Python program that accepts a string from the user. Your program should create a new string by shifting the string one position to the left.

```python
string = input("Enter a string: ")
new_string = string[1:] + string[0]
print("Shifted string:", new_string)
```

6. Write a program that asks the user to input his or her full name and print its initials. Assuming the user always types first name, middle name, and last name and does not include any unnecessary spaces.

```python
full_name = input("Enter your full name: ")
name_parts = full_name.split()
initials = ''.join([part[0].upper() for part in name_parts])
print("Initials:", initials)
```
