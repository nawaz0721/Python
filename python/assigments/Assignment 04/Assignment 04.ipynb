### List Assignment

1. Print Alternate Elements of List

```python
input_str = input("Enter elements of the list separated by spaces: ")
user_list = input_str.split()

print("Alternate elements of the list:")
for i in range(0, len(user_list), 2):
    print(user_list[i], end=" ")
print()
```

2. Reverse List Without reverse() Method

```python
input_str = input("Enter elements of the list separated by spaces: ")
user_list = input_str.split()

reversed_list = []
for i in range(len(user_list)-1, -1, -1):
    reversed_list.append(user_list[i])

print("Original list:", user_list)
print("Reversed list:", reversed_list)
```

3. Find Largest Number Without max()

```python
input_str = input("Enter numbers separated by spaces: ")
numbers = [int(x) for x in input_str.split()]

largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num

print("List:", numbers)
print("Largest number:", largest)
```

4. Rotate List Elements

```python
input_str = input("Enter elements of the list separated by spaces: ")
user_list = input_str.split()

if len(user_list) > 0:
    last_element = user_list[-1]
    for i in range(len(user_list)-1, 0, -1):
        user_list[i] = user_list[i-1]
    user_list[0] = last_element

print("Rotated list:", user_list)
```

5. Delete Word from String

```python
input_string = input("Enter a string: ")
word_to_delete = input("Enter the word to delete: ")

words = input_string.split()
new_words = []
for word in words:
    if word != word_to_delete:
        new_words.append(word)

result = ' '.join(new_words)
print("Original string:", input_string)
print("String after deletion:", result)
```

6. Date Format Conversion

```python
date_input = input("Enter date in mm/dd/yyyy format: ")

month, day, year = date_input.split('/')

months = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

month_name = months[int(month) - 1]
formatted_date = f"{month_name} {int(day)}, {year}"
print("Formatted date:", formatted_date)
```

7. Capitalize Each Word

```python
input_string = input("Enter a string: ")

words = input_string.split()
capitalized_words = []
for word in words:
    capitalized_word = word[0].upper() + word[1:].lower()
    capitalized_words.append(capitalized_word)

result = ' '.join(capitalized_words)
print("Original string:", input_string)
print("Capitalized string:", result)
```

8. Sum of Each Row in Matrix

```python
m = int(input("Enter number of rows: "))
n = int(input("Enter number of columns: "))

matrix = []

print(f"Enter {m}x{n} matrix:")
for i in range(m):
    row = []
    print(f"Row {i+1}:")
    for j in range(n):
        element = int(input(f"Enter element [{i+1}][{j+1}]: "))
        row.append(element)
    matrix.append(row)

print("\nMatrix:")
for row in matrix:
    print(row)

print("\nRow sums:")
for i in range(m):
    row_sum = 0
    for j in range(n):
        row_sum += matrix[i][j]
    print(f"Row {i+1} sum: {row_sum}")
    
```

9- Write  a program to add two matrices of size n x m

```python
##### User-friendly matrix addition with one-by-one input

n = int(input("Enter number of rows: "))
m = int(input("Enter number of columns: "))

matrix1 = []
matrix2 = []
result = []

print("\nEnter elements for Matrix 1:")
for i in range(n):
    row = []
    print(f"Row {i+1}:")
    for j in range(m):
        element = int(input(f"  Column {j+1}: "))
        row.append(element)
    matrix1.append(row)

print("\nEnter elements for Matrix 2:")
for i in range(n):
    row = []
    print(f"Row {i+1}:")
    for j in range(m):
        element = int(input(f"  Column {j+1}: "))
        row.append(element)
    matrix2.append(row)

# Add matrices
for i in range(n):
    row = []
    for j in range(m):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)

# Display results
print("\n" + "="*50)
print("MATRIX ADDITION RESULT")
print("="*50)

print("\nMatrix 1:")
for i in range(n):
    for j in range(m):
        print(f"{matrix1[i][j]:4}", end=" ")
    print()

print("\nMatrix 2:")
for i in range(n):
    for j in range(m):
        print(f"{matrix2[i][j]:4}", end=" ")
    print()

print("\nResult (Matrix 1 + Matrix 2):")
for i in range(n):
    for j in range(m):
        print(f"{result[i][j]:4}", end=" ")
    print()
```

10- Write a program to multiple two matrices

```python
n = int(input("Enter number of rows: "))
m = int(input("Enter number of columns: "))

matrix1 = []
matrix2 = []
result = []

print("\nEnter elements for Matrix 1:")
for i in range(n):
    row = []
    print(f"Row {i+1}:")
    for j in range(m):
        element = int(input(f"  Column {j+1}: "))
        row.append(element)
    matrix1.append(row)

print("\nEnter elements for Matrix 2:")
for i in range(n):
    row = []
    print(f"Row {i+1}:")
    for j in range(m):
        element = int(input(f"  Column {j+1}: "))
        row.append(element)
    matrix2.append(row)

# multiple matrices
for i in range(n):
    row = []
    for j in range(m):
        row.append(matrix1[i][j] * matrix2[i][j])
    result.append(row)

# Display results
print("\n" + "="*50)
print("MATRIX MULTIPLE RESULT")
print("="*50)

print("\nMatrix 1:")
for i in range(n):
    for j in range(m):
        print(f"{matrix1[i][j]:4}", end=" ")
    print()

print("\nMatrix 2:")
for i in range(n):
    for j in range(m):
        print(f"{matrix2[i][j]:4}", end=" ")
    print()

print("\nResult (Matrix 1 * Matrix 2):")
for i in range(n):
    for j in range(m):
        print(f"{result[i][j]:4}", end=" ")
    print()
```
