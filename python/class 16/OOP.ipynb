- Ploymorphism

```python
class Animal:
    def speak(self):
        return "Animal Speaks"

class Dog(Animal):
    def speak(self):
        return "Woof!"
        
class Cat(Animal):
    def speak(self):
        return "Meow!"

# create instances
dog = Dog()
cat = Cat()

# call the speak method on different objects
print(dog.speak())
print(cat.speak())
```

```python
class SSUET():

    def public_method(self):
        return "Public Method"

    def _protected_method(self):
        return "protected Method"

    def __private(self):
        return "Private Method"
```

```python
obj = SSUET()

print(obj.public_method())
print(obj._protected_method())
print(obj._SSUET__private())
```

- Use encapsulation for calss i.e StudentLogin():

- Create one private method i.e name __dbconectivity() that will vaildate you username and password.  if both are valid then should print "valid user" otherwise "not valid"
- Create one public methods i:e., update password & it should update your private password

```python
class Studentlogin():

    def __init__(self):
        self.__user_name = "Ahmed Nawaz"
        self.__password = "123456"

    def __dbconectivity(self, username, password):
        if (self.__user_name == username and self.__password == password):
            print("Valid User")
        else:
            print("Not valid")

    def login(self, username, password):
        self.__dbconectivity(username, password)

    def update_Password(self, password):
        self.__password = password
        print("Passowrd Updated Successfuly!")

        
```

### Data Files

- Read  --> 'r'
- Write --> 'w'
- Append --> 'a'
- Read/Write --> 'wr'

```python
#Method 1 - we have to specifically close the file 

f = open('file.txt', 'w')
f.write("This is a content file")
f.close()
```

```python
# Method 2 - file will close automatically (Recommended Method)

with open('file.txt' , 'w') as f:
    f.write("Now file will automatically close ")
```

```python
with open('file.txt' , 'a') as f:
    f.write("Now This line append in this file  \nThis will get appende on next line")
    f.writelines("This is a new")
```

```python
with open('file.txt' , 'r') as f:
    file_content = f.read()
    print(file_content)
```
