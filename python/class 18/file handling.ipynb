### CSV File Handling

```python
import csv
```

```python
competitions = []
with open('./competition.csv' , 'r') as f:
    content = csv.reader(f)
    print(content)
    for line in content:
        competitions.append(line)
        print(line)
    
```

```python
competitions
```

```python
year = []
event = []
winner = []

for x in competitions[1:]:
    year.append(x[0])
    event.append(x[1])
    winner.append(x[2])
```

```python
winner
```

```python
with open('./competition.csv' , mode= 'a' , newline= '') as f:
    content = csv.writer( f ,delimiter=",")
    content.writerows([["2025", "Asia Cup","Pakistan"],
                      ["2024", "FIFA","Argentina"]])
```
