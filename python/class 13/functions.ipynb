### Function in Function

```python
def factorial(n):
    if(n==0 or n==1):
        return 1
    else:
        return(n*factorial(n-1))
print(factorial(5))
```

### While Loop

Task: Research about while loop syntax and write a code of finding factorial using while loop

```python
factorial = 1
n = int(input("Enter Your Number"))

while(n>0):
    factorial*=n
    n-=1
print(factorial)
```

Every class have two things
 
  1. Characteristics --> Attributes
  2. Behaviours/Actions --> Methods
 
eg:
- Characteristics eg: color,  model, make_year
- Behaviours eg: start, end

```python
class cars():
  def __init__(self, color, model, make_year): #contructor
      self.color = color
      self.model = model
      self.make_year = make_year
      
  def start(self):    #method
      """
      Start belong to cars
      arg: no argument
      return: no return
      """
      print("car can start")
  def stop(self):    #method 
      """
      Stop belong to cars
      arg: no argument
      return: no return
      """
      print("car can stop")
```

```python
civic = cars('Black', 'Honda', 2025)
```

```python
print(type(civic))
```

```python
civic.start()
```

```python
civic.stop()
```

```python
civic.color, civic.model, civic.make_year
```

```python
class human():
  def __init__(self, age, year_of_birth): #contructor
      self.age = age
      self.year_of_birth = year_of_birth
      
  def coding(self):    #method
      """
      Codeing belongs to human
      arg: no argument
      return: no return
      """
      print("Good human is also a coder")
      
  def running(self):    #method 
      """
      running belong to human
      arg: no argument
      return: no return
      """
      print("Running is good for health")
```

```python
Ahmed = human(23, 2002)
```

```python
Ahmed.age
```

```python
Ahmed.year_of_birth
```
