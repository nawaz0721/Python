```python
class Parents():
    def __init__(self, eye_color, hair_color): #Constructor
        self.eye_color = eye_color #Attributes
        self.hair_color = hair_color

    def can_paint(self): #Methods
        print(f"Have ability to draw/paint arts")
```

```python
par_obj = Parents("Brown", "Brown")
#------------------------------
print(par_obj.eye_color)
print(par_obj.hair_color)
par_obj.can_paint()
```

```python
class Child(Parents):
    pass
```

```python
child_obj = Child("Brown", "Black")
#------------------------------
print(child_obj.eye_color)
print(child_obj.hair_color)
child_obj.can_paint()
```

```python
class child(Parents):

    def __init__(self, eye_color, hair_color, skin_color):
        super().__init__(eye_color, hair_color)
        self.skin_color = skin_color

    def can_drive(self):
        print(f"Child can driver")
```

```python
child_obj = child("Brown", "Black", "Brown")
#------------------------------
print(child_obj.eye_color)
print(child_obj.hair_color)
print(child_obj.skin_color)
child_obj.can_drive()
```

## Task:
 - Create a class with name "Father" which have two attributes and one method.
 - Create a class with name "Mother" which have one attribute and two methods.
 - Create a class with name "Child" that inherit characteris of "Father" and "Mother" but should have aleast two more attribute and methods other than "father class"

```python
class Father():
    def __init__(self, eye_color, hair_color): #Constructor
        self.eye_color = eye_color #Attributes
        self.hair_color = hair_color

    def can_drive(self): #Methods
        print(f"Can Drive ")
```

```python
class Mother():
    def __init__(self, skin_color): #Constructor
         self.skin_color = skin_color #Attributes

    def can_cook(self): #Methods
        print(f"Can cook for us")

    def can_teach(self):
        print(f"Can teach students")
```

```python
class childboy(Father, Mother):

    def __init__(self, eye_color, hair_color, skin_color, self_drive):
        Father.__init__(self,eye_color, hair_color)
        Mother.__init__(self, skin_color)
        self.self_drive = self_drive

    def can_drive(self):
        print(f"Child can driver")
```

```python
childboy_obj = childboy("Brown", "Black", "Brown", "Yes")
#------------------------------
print(childboy_obj.eye_color)
print(childboy_obj.hair_color)
print(childboy_obj.skin_color)
print(childboy_obj.self_drive)
child_obj.can_drive()
```
