## Lambda Function

- Without name
- Never been called beofre or after
- No return keyword

```python
(lambda x,y : x+y)(8,9)
```

```python
fun = lambda x,y : x*y
```

```python
fun(8,8)
```

```python
f = lambda n: 1 if n==1 else n*f(n-1)
```

```python
f(8)
```

## map(function , iterable)

```python
list(map(lambda x: (x, len(x)), ["Ali", "Ahmed", "Sir Qasim"]))
```

```python
dict(map(lambda x: (x, len(x)), ["Ali", "Ahmed", "Sir Qasim"]))
```

```python
set(map(lambda x: (x, len(x)), ["Ali", "Ahmed", "Sir Qasim"]))
```

```python
sum(map(lambda x: x**3, range(1,8,2)))
```

```python
any(map(lambda x : x>3, range(1,8)))
```

```python
all(map(lambda x : x>3, range(1,8)))
```

```python
list(map(lambda x : x>3, range(1,8)))
```

```python
list(filter(lambda x: x>3, range(1,9)))
```

```python
dict(filter(lambda x: x[1]>15, [("ahmed",25),("Hashim",20)]))
```

```python
# lambda <parameter> : <expression> (true if else false)
criteria =  lambda marks: "Pass" if marks > 50 else "Fail"
```

```python
criteria(55)
```

### Classes

```python
class saylani():
    helpline = '0900078601' # static
    obj_count = 0 #dynamic

    def __init__(self, campus, staff_count, course): #constructor
        self.campus = campus # attribute
        self.staff_count = staff_count
        self.course = course
        saylani.obj_count += 1

    def detail(self, courses): #methods
        print(f"{self.campus} have {self.staff_count} staff and courses are {self.course}")
        course = map(lambda x: x,courses)
        print(*course)
        print(f"courses: {courses}")
        # print(type(course))
        
    @staticmethod
    def facilaties():
        print("medical also")
```

```python
za = saylani("ZA IT Park", 10 , 5) #object initialization
```

```python
za.detail(["AI", "DE" , "WEB"])
```
